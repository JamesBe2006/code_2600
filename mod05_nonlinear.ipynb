# Module 05: Linear Regression with Nonlinear Effects

```python
# packages
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.pyplot import subplots
import statsmodels.api as sm
from sklearn.model_selection import train_test_split 
from ISLP import load_data
from ISLP.models import (ModelSpec as MS, summarize)

# set seed
seed = 4721
```

### We'll use the _Bikeshare_ data from ISLP for this activity. The metadata for _Bikeshare_ can be found [here](https://intro-stat-learning.github.io/ISLP/datasets/Bkeshare.html).

```python
# Load the data
Bikeshare = load_data('Bikeshare')
```

```python
# List the columns and their types
Bikeshare.dtypes
```

### We will predict the `bikers` column, which is the total of casual and registered bikers. 
Then we can create a matrix of potential independent variables (X) and a dependent variable vector (y).

```python
indep_vars = ['season', 'mnth', 'day', 'hr', 
              'holiday', 'weekday', 'workingday', 
              'weathersit', 'temp', 'atemp', 'hum', 'windspeed']

X = Bikeshare[indep_vars]
y = Bikeshare['bikers']
```

### Before doing any other analyses, let's create training and test sets.

```python
X_train, X_test, y_train, y_test, Train, Test = train_test_split(X, y, Bikeshare, 
                                                                 random_state = seed, 
                                                                 test_size = 0.25, 
                                                                 shuffle = True)
                             
```

### We can first summarize the variables in the training data.

```python
Train.describe()
```

### Then we can create plots for each potential independent variable with `bikers` on training data.

```python
# Initialize the plots before drawing them
nrows = 4
ncols = 3
figsize = (5*nrows, 5*ncols)

fig, axes = subplots(nrows=nrows,
                     ncols=ncols,
                     figsize=figsize)

# Assign a grid location to each index
def range_to_grid(i, nrows, ncols):
    x=[]
    y=[]
    for n in range(nrows*ncols):
        x.append(n // ncols)
        y.append(n % ncols)
        # print(n,x[n],y[n]) # for testing this function
    return x[i],y[i]

# Plot the variables
for j in range(len(X_train.columns)):
    # print(range_to_grid(j,nrows,ncols)[0], range_to_grid(j,nrows,ncols)[1]) # testing
    axes[range_to_grid(j,nrows,ncols)[0],
         range_to_grid(j,nrows,ncols)[1]].plot(X_train.iloc[:,j], y_train, 'o')
    axes[range_to_grid(j,nrows,ncols)[0],
         range_to_grid(j,nrows,ncols)[1]].set_xlabel(X_train.columns[j])
```

### Since we see that the `hr` variable is periodic (with period 24), we will use trigonometric transforms as our spline basis functions

```python
# Data preprocessing
Train['hr_sin'] = round(np.sin(2*np.pi/24*Train['hr'].astype('int')),2)
Test['hr_sin'] = round(np.sin(2*np.pi/24*Test['hr'].astype('int')),2)
X_train['hr_sin'] = round(np.sin(2*np.pi/24*X_train['hr'].astype('int')),2)
X_test['hr_sin'] = round(np.sin(2*np.pi/24*X_test['hr'].astype('int')),2)

Train['hr_cos'] = round(np.cos(2*np.pi/24*Train['hr'].astype('int')),2)
Test['hr_cos'] = round(np.cos(2*np.pi/24*Test['hr'].astype('int')),2)
X_train['hr_cos'] = round(np.cos(2*np.pi/24*X_train['hr'].astype('int')),2)
X_test['hr_cos'] = round(np.cos(2*np.pi/24*X_test['hr'].astype('int')),2)

# spot-check this
Train.head()
```

## First Linear Regression Model with a spline

### We can create a column to represent the intercept in our model. 

Some packages do this automatically, but we will leverage the ISLP version, which does not.

```python
X_train['intercept'] = np.ones(X_train.shape[0])
X_test['intercept'] = np.ones(X_test.shape[0])
```

### We'll build a regression model to predict `bikers` that uses a straight line for `temp` and a spline with trig basis functions for `hr`.

```python
model_4 = sm.OLS(y_train, X_train[['intercept', 'temp', 'hr_sin', 'hr_cos']])
results_4 = model_4.fit()
summarize(results_4)
```

### Let's create a plot through `hr` to visualize the resulting spline.

```python
# Choose a representative temperature
temp_value = X_train["temp"].mean()

# Smooth range of hours
hr_grid = np.linspace(0, 24, 200)

# Construct the design matrix for prediction
X_grid = pd.DataFrame({
    "intercept": 1,
    "temp": temp_value,
    "hr_sin": np.sin(2 * np.pi * hr_grid / 24),
    "hr_cos": np.cos(2 * np.pi * hr_grid / 24)
})

# Predicted target
y_pred = results_4.predict(X_grid)

plt.figure(figsize=(9, 5))

# Actual observations
plt.scatter(
    X_train["hr"],
    y_train,
    alpha=0.15,
    s=15,
    label="Observed"
)

# Fitted curve
plt.plot(
    hr_grid,
    y_pred,
    color="red",
    linewidth=3,
    label=f"Fitted curve (temp = {temp_value:.1f})"
)

plt.xlabel("Hour")
plt.ylabel("Target")
plt.title("Trig spline fit for hour")
plt.xticks(range(0, 25, 2))
plt.grid(alpha=0.3)
plt.legend()
plt.show()
```

### We will evaluate this model *three ways*: Using R^2, using MSE, and then visually. 

For the last two, we will compare on _Train_ and on _Test_.

```python
# R^2 on training
results_4.rsquared
```

```python
# Create helper functions for computing the mean squared error

def predict(X, model):
    # the built-in get_prediction tool returns an array, so we need to convert to a dataframe
    predictions_df = pd.DataFrame(model.get_prediction(X).predicted, columns=['y_hat'], index=X.index)
    return predictions_df['y_hat']

def mse(y, y_hat):
    # calculate the residual error for each individual record
    resid = y - y_hat
    # square the residual (hence "squared error")
    sq_resid = resid**2
    # calculate the sum of squared errors
    SSR = sum(sq_resid)
    # divide by the number of records to get the mean squared error
    MSE = SSR / y.shape[0]
    return MSE
```

```python
predictions_train_4 = predict(X_train[['intercept', 'temp', 'hr_sin', 'hr_cos']], results_4)
print('MSE train: ', mse(y_train, predictions_train_4))

predictions_test_4 = predict(X_test[['intercept', 'temp', 'hr_sin', 'hr_cos']], results_4)
print('MSE test: ', mse(y_test, predictions_test_4))
```

### Here is a helpful tool for evaluating our model(s). 

Since the variables we're using aren't always being fit with straight lines, it's helpful to plot the average _actual_ y-value to the average _predicted_ y-value for each modeled feature.

**Note**: It is **not** recommended to leverage this tool on test data. This can quickly lead to unintentional overfitting on the test set, which could lead to a poor model on new data.

```python
def plot_avg_actual_vs_predicted_by_feature(X, y, y_pred, feature, target_name="Target"):
    df = pd.DataFrame({
        feature: X[feature],
        "actual": y,
        "predicted": y_pred
    })

    grouped = df.groupby(feature).mean().reset_index()

    plt.figure()
    plt.plot(grouped[feature], grouped["actual"], label="Average Actual")
    plt.plot(grouped[feature], grouped["predicted"], label="Average Predicted")
    plt.xlabel(feature)
    plt.ylabel(target_name)
    plt.legend()
    plt.show()
```

```python
plot_avg_actual_vs_predicted_by_feature(X=X_train, 
                                        y=y_train, 
                                        y_pred=predictions_train_4, 
                                        feature='temp', 
                                        target_name="Bikers"
                                        )
```

```python
plot_avg_actual_vs_predicted_by_feature(X=X_train, 
                                        y=y_train, 
                                        y_pred=predictions_train_4, 
                                        feature='hr', 
                                        target_name="Bikers"
                                        )
```

### We are seeing that there are spikes in bike rentals at rushhour times: 7am, 8am, 5pm, and 6pm. Let's incorporate a categorical variable to account for these.

```python
results_5.rsquared
```

```python
from datetime import datetime

# Data pre-processing (on Train, Test, X_train, and X_test)
Train['light'] = np.where(Train['weathersit'] == 'light rain/snow', 1, 0)
Test['light'] = np.where(Test['weathersit'] == 'light rain/snow', 1, 0)
X_train['heavy'] = np.where(X_train['weathersit'] == 'heavy rain/snow', 1, 0)
X_test['heavy'] = np.where(X_test['weathersit'] == 'heavy rain/snow', 1, 0)

Train['light'] = np.where(Train['weathersit'] == 'light rain/snow', 1, 0)
Test['light'] = np.where(Test['weathersit'] == 'light rain/snow', 1, 0)
X_train['light'] = np.where(X_train['weathersit'] == 'light rain/snow', 1, 0)
X_test['light'] = np.where(X_test['weathersit'] == 'light rain/snow', 1, 0)

Train['clear'] = np.where(Train['weathersit'] == 'clear', 1, 0)
Test['clear'] = np.where(Test['weathersit'] == 'clear', 1, 0)
X_train['clear'] = np.where(X_train['weathersit'] == 'clear', 1, 0)
X_test['clear'] = np.where(X_test['weathersit'] == 'clear', 1, 0)

Train['overcast'] = np.where(Train['weathersit'] == 'cloudy/misty', 1, 0)
Test['overcast'] = np.where(Test['weathersit'] == 'cloudy/misty', 1, 0)
X_train['overcast'] = np.where(X_train['weathersit'] == 'cloudy/misty', 1, 0)
X_test['overcast'] = np.where(X_test['weathersit'] == 'cloudy/misty', 1, 0)

Train['sin_day'] = np.sin(2*np.pi/365*Train['day'].astype('int'))
Test['sin_day'] = np.sin(2*np.pi/365*Test['day'].astype('int'))
X_train['sin_day'] = np.sin(2*np.pi/365*X_train['day'].astype('int'))
X_test['sin_day'] = np.sin(2*np.pi/365*X_test['day'].astype('int'))

Train['cos_day'] = np.cos(2*np.pi/365*Train['day'].astype('int'))
Test['cos_day'] = np.cos(2*np.pi/365*Test['day'].astype('int'))
X_train['cos_day'] = np.cos(2*np.pi/365*X_train['day'].astype('int'))
X_test['cos_day'] = np.cos(2*np.pi/365*X_test['day'].astype('int'))

Train['conditions'] = Train['temp'] * Train['hum'] * Train['windspeed']
Test['conditions'] = Test['temp'] * Test['hum'] * Test['windspeed']
X_train['conditions'] = X_train['temp'] * X_train['hum'] * X_train['windspeed']
X_test['conditions'] = X_test['temp'] * X_test['hum'] * X_test['windspeed']

Train['month_num'] = Train['mnth'].apply(lambda x: datetime.strptime(x[:3], "%b").month)
Test['month_num'] = Test['mnth'].apply(lambda x: datetime.strptime(x[:3], "%b").month)
X_train["month_num"] = X_train["mnth"].apply(lambda x: datetime.strptime(x[:3], "%b").month)
X_test["month_num"] = X_test["mnth"].apply(lambda x: datetime.strptime(x[:3], "%b").month)

Train['sin_month'] = np.sin(2*np.pi/12*Train['month_num'].astype('int'))
Test['sin_month'] = np.sin(2*np.pi/12*Test['month_num'].astype('int'))
X_train['sin_month'] = np.sin(2*np.pi/12*X_train['month_num'].astype('int'))
X_test['sin_month'] = np.sin(2*np.pi/12*X_test['month_num'].astype('int'))

Train['cos_month'] = np.sin(2*np.pi/12*Train['month_num'].astype('int'))
Test['cos_month'] = np.cos(2*np.pi/12*Test['month_num'].astype('int'))
X_train['cos_month'] = np.cos(2*np.pi/12*X_train['month_num'].astype('int'))
X_test['cos_month'] = np.cos(2*np.pi/12*X_test['month_num'].astype('int'))

Train['sin_weekday'] = np.sin(2*np.pi/7*Train['weekday'].astype('int'))
Test['sin_weekday'] = np.sin(2*np.pi/7*Test['weekday'].astype('int'))
X_train['sin_weekday'] = np.sin(2*np.pi/7*X_train['weekday'].astype('int'))
X_test['sin_weekday'] = np.sin(2*np.pi/7*X_test['weekday'].astype('int'))

Train['cos_weekday'] = np.cos(2*np.pi/7*Train['weekday'].astype('int'))
Test['cos_weekday'] = np.cos(2*np.pi/7*Test['weekday'].astype('int'))
X_train['cos_weekday'] = np.cos(2*np.pi/7*X_train['weekday'].astype('int'))
X_test['cos_weekday'] = np.cos(2*np.pi/7*X_test['weekday'].astype('int'))

Train['sin_season'] = np.sin(2*np.pi/4*Train['season'].astype('int'))
Test['sin_season'] = np.sin(2*np.pi/4*Test['season'].astype('int'))
X_train['sin_season'] = np.sin(2*np.pi/4*X_train['season'].astype('int'))
X_test['sin_season'] = np.sin(2*np.pi/4*X_test['season'].astype('int'))

Train['cos_season'] = np.cos(2*np.pi/4*Train['season'].astype('int'))
Test['cos_season'] = np.cos(2*np.pi/4*Test['season'].astype('int'))
X_train['cos_season'] = np.cos(2*np.pi/4*X_train['season'].astype('int'))
X_test['cos_season'] = np.cos(2*np.pi/4*X_test['season'].astype('int'))

Train['temp_x_hum'] = Train['temp'] * Train['hum']
Test['temp_x_hum'] = Test['temp'] * Test['hum']
X_train['temp_x_hum'] = X_train['temp'] * X_train['hum']
X_test['temp_x_hum'] = X_test['temp'] * X_test['hum']

Train['temp_x_windspeed'] = Train['temp'] * Train['windspeed']
Test['temp_x_windspeed'] = Test['temp'] * Test['windspeed']
X_train['temp_x_windspeed'] = X_train['temp'] * X_train['windspeed']
X_test['temp_x_windspeed'] = X_test['temp'] * X_test['windspeed']

Train['hum_x_windspeed'] = Train['hum'] * Train['windspeed']
Test['hum_x_windspeed'] = Test['hum'] * Test['windspeed']
X_train['hum_x_windspeed'] = X_train['hum'] * X_train['windspeed']
X_test['hum_x_windspeed'] = X_test['hum'] * X_test['windspeed']

Train['rushhour_x_workingday'] = Train['rushhour'] * Train['workingday']
Test['rushhour_x_workingday'] = Test['rushhour'] * Test['workingday']
X_train['rushhour_x_workingday'] = X_train['rushhour'] * X_train['workingday']
X_test['rushhour_x_workingday'] = X_test['rushhour'] * X_test['workingday']
```
