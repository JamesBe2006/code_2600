### Create a basic random forest model on the training set.

```python
rf_basic = RFC(max_features=len(indep_vars),
               criterion='log_loss',
               n_estimators=50,
               max_depth=10,
               min_samples_leaf=1,
               bootstrap=True,
               oob_score=True,
               random_state=seed)
rf_basic.fit(X_train, y_train)
```

### We'll use this function to quickly calculate the log loss of a model.

```python
def log_loss(y_true, y_prob, eps=1e-15):
    y_true = np.array(y_true)
    y_prob = np.array(y_prob)
    
    # avoid log(0)
    y_prob = np.clip(y_prob, eps, 1 - eps)
    
    ll = -np.mean(y_true * np.log(y_prob) + (1 - y_true) * np.log(1 - y_prob))
    return ll
```

### Compute the log loss on the training set.

```python
y_hat_basic_train = rf_basic.predict_proba(X_train)[:,1]
print('train log loss:', log_loss(y_train, y_hat_basic_train))
```

### Calculate the out-of-bag error (bootstrapping) on training data to estimate the log loss on the test set.

```python
y_hat_basic_oob = rf_basic.oob_decision_function_[:,1]
print('oob log loss:  ', log_loss(y_train, y_hat_basic_oob))
```

### Compute the log loss on the test set.

```python
y_hat_basic_test = rf_basic.predict_proba(X_test)[:,1]
print('test log loss: ', log_loss(y_test, y_hat_basic_test))
```

### (2) How is “log loss” different from the log likelihood we defined in class?

Log loss is a transformation of the log likelihood function, which approached 0 as the prediction is more accurate. The log likelihood function take the likelihood function and takes the log of every f(x), allowing us to add these together instead of multiplying. The goal of this function is to maximize its value, which shows the parameters that make the observed outcome the most likely. The log loss multiplies this by -(1/N). Since logs of probabilities between 0 and 1 are negative, the -(1/N) corrects this to positive. Maximizing the log likelihood will have the effect of minimizing the log loss, and the 1/N term allows us to standardize this metric across datasets of different sizes.

### (3) Explain how we know the basic model is overfit before attempting to build any other models.

By lookinng at the train log loss as compared to the out-of-bag (OOB) log loss and test log loss, we can diagnose a big difference as overfitting. The OOB and test are very close (both aroung 0.66), allowing OOB to be a good representation of the test log loss. The large gap, 0.46 for the test log loss as opposed to 0.66 for the OOB and test log loss, shows that the model memorized the train data and is performing poorly elsewhere.
